# NumPy Indexing, Slicing and 2D Arrays

## Learning Objectives
- Access individual array elements
- Extract portions of an array
- Understand 2D arrays and matrices
- Select rows, columns, and individual values

```python
import numpy as np

numbers = np.array([10,20,40,70,90,110,257])

print ("First Element:", numbers[0]) #Will print element present at index 0 means the very first element in the array
print ("Second Element:", numbers[1])
print ("Fourth Element:", numbers[3])
print ("Last Element:", numbers[6])
print ("last Element:", numbers[-1]) # will print last element but starts counting from reverse so, that's why -1 will print last element
print ("Second last Element:", numbers[-2])



# SLICING
# Array [10,20,40,70,90,110,257]

print ("First Three:", numbers[:3])  #Prints only first three numbers in the array

print ("Elements from index 2:", numbers[2:]) # will print all the elements staring from index No.02

print ("Middle Elements:", numbers[1:4]) # will print from index 1 and stop at index 4 means numbers will be print from index 1 to index 3

print ("Elements with gap of 2 steps:", numbers[0:7:2]) # It will start from index 0 (10), then move forward by 2 steps to print index 2 (40), then index 4 (90), and then index 6 (257).

print("Every second element:", numbers[::2]) # will print array elements with the gap of 2 in between the elements starting from index 0 to index 6 same as above print statement

print ("Copy of the Array:", numbers[::]) # will print all the elements of the Array or we can say copy of the Array

```

```python
# 2D arrays (matrices)

matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Matrix:\n", matrix)
print("Shape:", matrix.shape)
print("First row:", matrix[0])
print("Second row, third column:", matrix[1, 2])
print("First column:", matrix[:, 0])
print("Last two columns:\n", matrix[:, 1:])
```
