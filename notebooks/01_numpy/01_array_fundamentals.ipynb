# NumPy — Array Fundamentals

## Learning Objectives
- Understand NumPy arrays
- Explore array shape and dimensions
- Inspect data types
- Perform basic numerical calculations

## Practice
Build and analyze a numerical array using NumPy.

```python
import numpy as np 

numbers = np.array([10,40,60,80,30,90])

print("Array:", numbers) 

print("Shape:", numbers.shape)


arr_1d = np.array([10, 20, 30, 40])
print(arr_1d.shape)  


# 2. Two-Dimensional Array (Matrix)
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print(arr_2d.shape)  

# 3. Three-Dimensional Array
arr_3d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(arr_3d.shape)  


print("Dimensions:", numbers.ndim)
print("Dimensions:", arr_1d.ndim)
print("Dimensions:", arr_2d.ndim)
print("Dimensions:", arr_3d.ndim)



print("Data type", numbers.dtype)
print("Number Of Elements:", numbers.size)
print("Mean", np.mean(numbers))
print("Maximum:", np.max(numbers))
print("Minimum:", np.min(numbers) )
```
